# 2026 Rainfall Chart (Jan–Jun)

Monthly rainfall totals and departure from normal.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "month": ["Jan", "Feb", "March", "April", "May", "June"],
    "rainfall_in": [5.1, 8.1, 22.3, 6.1, 6.1, 7.0],
    "departure_in": [-0.1, 3.0, 15.4, 1.0, 2.0, 3.4],
}

df = pd.DataFrame(data)
df
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(df["month"], df["rainfall_in"], color="#4C72B0")

for bar, rain, dep in zip(bars, df["rainfall_in"], df["departure_in"]):
    sign = "+" if dep >= 0 else ""
    label = f"{rain:.1f} in\n({sign}{dep:.1f} in)"
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.3,
        label,
        ha="center",
        va="bottom",
        fontsize=9,
    )

ax.set_ylabel("Rainfall (in)")
ax.set_title("2026 Monthly Rainfall (with departure from normal)")
ax.set_ylim(0, df["rainfall_in"].max() + 4)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

colors = ["#C44E52" if d < 0 else "#55A868" for d in df["departure_in"]]
ax.bar(df["month"], df["departure_in"], color=colors)
ax.axhline(0, color="black", linewidth=0.8)

ax.set_ylabel("Departure from normal (in)")
ax.set_title("2026 Monthly Rainfall Departure from Normal")

plt.tight_layout()
plt.show()
```
